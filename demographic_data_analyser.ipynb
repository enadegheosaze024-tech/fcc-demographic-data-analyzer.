```python
import pandas as pd
df = pd.read_csv('adult.data.csv')
df.columns = df.columns.str.strip()
df.index = df.index + 1
df.head()
```

```python
df['race'].value_counts()
```

```python
# Create a 'filter' for Males..This filter says "Find every row where the sex is exactly "male"
male = df['sex'] == 'Male'

# Use that filter to create a new, smaller table of just men
men_df = df[male]

# Check the first few rows to make sure no women are in there
men_df.head()
```

```python
#Calculate the average of the age of the men
Average_age_of_men =round( men_df['age'].mean(),1)
print('Average age of men:', Average_age_of_men)
```

```python
# Part: Count rows where education is Bachelors
bachelors_count = len(df[df['education'] == 'Bachelors'])

# Whole: Count all rows
total_count = len(df)

# Result
print("Percentage of bachelors:",round((bachelors_count / total_count) * 100,1))
```

```python
#Create a list of the advanced degrees
Advanced_degrees = ['Bachelors', 'Masters', 'Doctorates']
#Filter those with advanced degrees as "higher education" and thos without it as 'lower education'
higher_education = df[df['education'].isin(Advanced_degrees)]
lower_education = df[~df['education'].isin(Advanced_degrees)]
print(f"High Ed rows: {len(higher_education)}")
print(f"Low Ed rows: {len(lower_education)}")
```

```python
# Count how many in the 'High Ed' group make >50K
high_ed_rich = len(higher_education[higher_education['salary'] == '>50K'])

#Calculate the percentage relative to the High Ed group only
perc_of_rich_high_ed = round((high_ed_rich / len(higher_education)) * 100,1)
print('Percentage of people with advanced education:',perc_of_rich_high_ed)
```

```python
# Use the .min() function to find the lowest value
min_hours = df['hours-per-week'].min()
print(min_hours)
```

```python
#Filter for only the people who work the minimum hours
num_min_workers = df[df['hours-per-week'] == min_hours]

#Count how many of THEM make >50K
rich_percentage_min_workers = len(num_min_workers[num_min_workers['salary'] == '>50K'])

# Calculate the percentage
rich_percentage = round((rich_percentage_min_workers / len(num_min_workers)) * 100 , 1)
print('rich percentage:',rich_percentage)
```

```python
#Total people per country
total_per_country = df['native-country'].value_counts()

#Rich people (>50K) per country
rich_per_country = df[df['salary'] == '>50K']['native-country'].value_counts()

#The Percentage (This creates a list of percentages for all countries)
country_percentages = (rich_per_country / total_per_country) * 100

#Find the highest name and the highest value
highest_earning_country = country_percentages.idxmax()
highest_earning_country_percentage = round(country_percentages.max(), 1)

print(f"Result: {highest_earning_country} at {highest_earning_country_percentage}%")
```

```python
# Filter for people in India AND earning >50K
india_rich = df[(df['native-country'] == 'India') & (df['salary'] == '>50K')]

#Find the most common value in the 'occupation' column
top_IN_occupation = india_rich['occupation'].value_counts().idxmax()

print(f"The top job in India for high earners is: {top_IN_occupation}")
```
